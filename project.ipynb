```python
import matplotlib.pyplot as plt
df = pd.read_csv("crimedata.csv")
df = df.rename(columns={
    'Crime Head Ipc (Indian Penal Code) Category': 'Category',
    'Crime Head Ipc (Indian Penal Code) Sub-Category': 'Sub_Category',
    'Incidence Of Ipc (Indian Penal Code) Crimes (UOM:Number), Scaling Factor:1': 'Incidence',
    'Victims Of Ipc (Indian Penal Code) Crimes (UOM:Number), Scaling Factor:1': 'Victims',
    'Ipc (Indian Penal Code) Crime Rate Per Lakh Person (UOM:PerLakh(100000)), Scaling Factor:1': 'Crime_Rate'
})
```

```python
df['Incidence'] = pd.to_numeric(df['Incidence'], errors='coerce')
df['Victims'] = pd.to_numeric(df['Victims'], errors='coerce')
```

```python
top_states = df.groupby('State')['Incidence'].sum().sort_values(ascending=False).head(10)

top_states.plot(kind='bar')
plt.title("Top 10 States by Crime")
plt.xlabel("State")
plt.ylabel("Incidence")
plt.xticks(rotation=45)
plt.show()
```

```python
yearly = df.groupby('Year')['Incidence'].sum()

yearly.plot(kind='line', marker='o')
plt.title("Year-wise Crime Trend")
plt.xlabel("Year")
plt.ylabel("Incidence")
plt.grid()
plt.show()
```

```python
df['Incidence'].plot(kind='hist', bins=20)
plt.title("Crime Distribution")
plt.xlabel("Incidence")
plt.show()
```

```python
plt.scatter(df['Incidence'], df['Victims'])
plt.title("Incidence vs Victims")
plt.xlabel("Incidence")
plt.ylabel("Victims")
plt.show()
```

```python
plt.style.use('ggplot')
```

```python
from sklearn.cluster import KMeans

X = df[['Incidence', 'Crime_Rate']]

kmeans = KMeans(n_clusters=3)
df['Cluster'] = kmeans.fit_predict(X)

print(df[['State', 'Cluster']].head())
```

```python
plt.figure(figsize=(7,7))

category = df.groupby('Category')['Incidence'].sum()

colors = ['gold','lightcoral','skyblue','lightgreen','pink','orange']

category.plot(kind='pie', autopct='%1.1f%%', colors=colors, shadow=True, startangle=140)

plt.title("Crime Category Distribution", fontsize=14, fontweight='bold')
plt.ylabel("")

plt.show()
```

```python
import seaborn as sns

sns.heatmap(df[['Incidence','Victims','Crime_Rate']].corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

X = df[['Year', 'Crime_Rate']]
y = df['Incidence']

model = RandomForestRegressor()
model.fit(X, y)

print(model.feature_importances_)
```

```python
from sklearn.cluster import KMeans

X = df[['Incidence', 'Crime_Rate']]

kmeans = KMeans(n_clusters=3)
df['Cluster'] = kmeans.fit_predict(X)

# GRAPH
plt.scatter(df['Incidence'], df['Crime_Rate'], c=df['Cluster'])
plt.title("Crime Clustering")
plt.xlabel("Incidence")
plt.ylabel("Crime Rate")
plt.show()
```

```python
wcss = []

for i in range(1, 6):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(df[['Incidence', 'Crime_Rate']])
    wcss.append(kmeans.inertia_)

plt.plot(range(1,6), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
from sklearn.decomposition import PCA

X = df[['Incidence','Crime_Rate','Victims']]

pca = PCA(n_components=2)
reduced = pca.fit_transform(X)

plt.scatter(reduced[:,0], reduced[:,1])
plt.title("PCA Visualization")
plt.show()
```

```python
plt.figure(figsize=(10,6))

for cat in df['Category'].unique()[:3]:  # limit for clarity
    subset = df[df['Category'] == cat]
    sns.kdeplot(subset['Incidence'], label=cat, fill=True, alpha=0.4)

plt.title("Crime Distribution by Category")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(x='Category', y='Crime_Rate', data=df)

plt.title("Crime Rate by Category", fontsize=14, fontweight='bold')
plt.xticks(rotation=45)

plt.show()
```
